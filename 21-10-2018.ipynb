# Polynomial Fitting

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(6)
x = np.linspace(0,1,20)
y = np.cos(x) + 0.1*np.random.rand(20)
p = np.poly1d(np.polyfit(x, y, 3))
print (p)

t = np.linspace(0,1,200)
plt.plot(x, y, '*', t, p(t), '-')
plt.xlabel('X-Values')
plt.ylabel('Y-Values')
plt.title("Curve Fitting")
plt.show()
```
