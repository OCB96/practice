Logistic Regression can be used to solve multi-class problems too. How?

P = (e^(w0+w1x1+w2x2))/(1+(e^(w0+w1x1+w2x2)))

P = Probability of belonging to class 1

1-P = Probability of NOT belonging to class 1

P/1-P = odds of belonging to class 1

ln(P/1-P) = w0+w1x1+w2x2 = a linear function

Loss function (mean squared error) = (1/N)(SUMMATION(P-Y)^2)
